```python
from rdkit import Chem
from rdkit.Chem import AllChem
import os


def add_hydrogens_to_mol(input_mol_file):
    """
    读取MOL文件，添加氢原子，并保存为新的MOL和XYZ文件
    
    参数:
        input_mol_file (str): 输入的MOL文件路径
    
    返回:
        tuple: (mol_output_path, xyz_output_path) 输出文件路径
    """
    
    # 检查输入文件是否存在
    if not os.path.exists(input_mol_file):
        raise FileNotFoundError(f"文件不存在: {input_mol_file}")
    
    # 读取MOL文件
    mol = Chem.MolFromMolFile(input_mol_file, removeHs=False)
    
    if mol is None:
        raise ValueError(f"无法读取MOL文件: {input_mol_file}")
    
    # 添加氢原子，并保持原有坐标
    mol_with_h = Chem.AddHs(mol, addCoords=True)
    
    # 生成输出文件名
    base_name = os.path.splitext(input_mol_file)[0]
    mol_output = f"{base_name}_addH.mol"
    xyz_output = f"{base_name}_addH.xyz"
    
    # 保存为MOL文件
    writer = Chem.SDWriter(mol_output)
    writer.write(mol_with_h)
    writer.close()
    
    # 保存为XYZ文件
    xyz_content = Chem.MolToXYZBlock(mol_with_h)
    with open(xyz_output, 'w') as f:
        f.write(xyz_content)
    
    print(f"补氢完成!")
    print(f"原始分子原子数: {mol.GetNumAtoms()}")
    print(f"补氢后原子数: {mol_with_h.GetNumAtoms()}")
    print(f"添加的氢原子数: {mol_with_h.GetNumAtoms() - mol.GetNumAtoms()}")
    print(f"MOL文件已保存至: {mol_output}")
    print(f"XYZ文件已保存至: {xyz_output}")
    
    return mol_output, xyz_output


```

```python

mol_files=glob.glob('./calculate_0129/*.mol')
# 使用示例
for input_file in mol_files:
    #input_file = "1767448938361_ald_from_ald-1_ami-1.mol"    
    try:
        mol_file, xyz_file = add_hydrogens_to_mol(input_file)
        print(f"\n处理成功!")
    except Exception as e:
        print(f"错误: {e}")
```
